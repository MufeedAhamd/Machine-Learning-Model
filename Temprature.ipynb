Real-Time Temperature Prediction 

```python
import pandas as pd 
import numpy as np
from sklearn.preprocessing import LabelEncoder , StandardScaler , OneHotEncoder
```

```python
df = pd.read_csv(r"C:\Users\Lenovo\Downloads\IndianWeatherRepository.csv")
```

```python
df1 =  df[['latitude', 'longitude','timezone' ,'temperature_celsius', 'condition_text','wind_kph', 'pressure_in','humidity','cloud']]
```

```python
df1
```

```python
print(df1.duplicated().sum())

# df1 = df1.drop_duplicates()
```

```python
X = df1.drop(columns = ["temperature_celsius"] , axis = 1)
y = df1["temperature_celsius"]
```

```python
cat = X.select_dtypes(include="object").columns
num = X.select_dtypes(exclude="object").columns
```

```python
from sklearn.compose import ColumnTransformer 
pre = ColumnTransformer(
    transformers=[
        ("cat"  , OneHotEncoder(handle_unknown="ignore"  , drop = "first"), cat),
        ("num" , StandardScaler() , num)
        
    ]
)
pre
```

```python
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor


# model = Pipeline(steps= [
#     ("preporcesser", pre),
#     ("linear" , LinearRegression())
# ])

from sklearn.pipeline import Pipeline

model = Pipeline([
    ('preprocessor', pre),
    ('regressor', RandomForestRegressor(
        n_estimators=200,
    max_depth=10,              # tree depth limit
    min_samples_split=10,      # split ke liye min samples
    min_samples_leaf=5,        # leaf me min data
    random_state=42
    ))
])
model
```

```python
from sklearn.model_selection import train_test_split

X_train , X_test , y_train , y_test = train_test_split(X,y , test_size=.2 ,random_state=42)
```

```python
from sklearn.metrics import r2_score
```

```python
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)
y_pred2 = model.predict(X_train)
```

```python
from sklearn.metrics import r2_score

y_pred = model.predict(X_test)

acc1 = r2_score(y_test, y_pred)   # ✅ correct order
acc2= r2_score(y_train, y_pred2)   # ✅ correct order
print("R2 Score (test):", acc1)
print("R2 Score (train):", acc2)
```

```python
import joblib 

joblib.dump(model, "weather_model.pkl")
```

```python
import joblib 
import streamlit as st

model = joblib.load("weather_model.pkl")

st.title("Weather Temperature Prediction")

#input
latitude = st.number_input("Latitude (north or south(°)) :")
longitude = st.number_input("Longitude (east or west(°)) :")

timezone = st.selectbox("Timezone", ["Asia/Kolkata"])

condition = st.selectbox("Condition", ["Sunny", "Cloudy", "Partly cloudy"])

wind_kph = st.number_input("Wind Speed (kph)")
pressure_in = st.number_input("Pressure (in)")
humidity = st.slider("Humidity", 0, 100)
cloud = st.slider("Cloud", 0, 100)

# predict
if st.button("Predict Temperature"):

    live = pd.DataFrame({
        'latitude':[latitude], 
        'longitude':[longitude], 
        'timezone':[timezone],
        'condition_text':[condition], 
        'wind_kph':[wind_kph], 
        'pressure_in':[pressure_in], 
        'humidity':[humidity], 
        'cloud':[cloud]
    })

    prediction = model.predict(live)

    st.success(f"🌡️ Predicted Temperature: {prediction[0]:.2f} °C")
```
